```python
import pandas as pd
import re


def read_txt_file(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()
            return content
    except FileNotFoundError:
        print(f"File not found: {file_path}")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None

def process_text_into_dataframe(content):
    # Split text using "Reviews" as the delimiter
    parts = content.split("Reviews")
    
    # Remove empty sections and strip whitespace
    parts = [part.strip() for part in parts if part.strip()]
    
    # Create the DataFrame
    df = pd.DataFrame(parts, columns=['Review'])
    
    # Add a sequence/index column starting from 6
    df['Sequence'] = range(6, 6 + len(df))
    
    # Reorder columns and sort by Sequence
    df = df[['Sequence', 'Review']].sort_values(by='Sequence').reset_index(drop=True)
    
    return df

# Replace with your target TXT file path
file_path = 'E:/EGM/editor/17-125.txt'
file_content = read_txt_file(file_path)

if file_content:
    df = process_text_into_dataframe(file_content)
    
    # Export DataFrame to CSV
    csv_file_path = 'E:/EGM/editor/6-125.csv'
    df.to_csv(csv_file_path, index=False, encoding='utf-8-sig')
    print(f"DataFrame has been saved to {csv_file_path}")
else:
    print("Failed to read the file.")
```
